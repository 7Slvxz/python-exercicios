```python
nome = input("Digite o nome do aluno: ")

# Responsável por armazenar o historico de disciplinas
historico = []
# Estrutura para repetir o cadastro de disciplinas
continuar = "s"
while continuar == "s":

    disciplina = input("Digite o nome da disciplina: ")

    soma_notas = 0
    soma_freq = 0

    # Estrutura de repetição para as 4 notas/frequências
    for i in range(1, 5):

        # Validação da Nota (Garante que seja entre 0 e 10)
        nota = -1
        while nota < 0 or nota > 10:
            nota = float(input(f"Digite a {i}ª nota (0-10): "))

            if nota < 0 or nota > 10:
                print("Valor inválido! Digite entre 0 e 10.")
        soma_notas += nota

        # Validação da Frequência (Garante que seja entre 0 e 100)
        freq = -1
        while freq < 0 or freq > 100:
            freq = int(input(f"Digite a {i}ª frequência (0-100): "))

            if freq < 0 or freq > 100:
                print("Valor inválido! Digite entre 0 e 100.")
        soma_freq += freq

    nota_media = soma_notas / 4
    frequencia_media = soma_freq / 4

    print(f"Aluno: {nome} | Disciplina: {disciplina}")
    print(f"Média Final: {nota_media:}")
    print(f"Frequência Média: {frequencia_media:}%")

    situacao = ""

    if nota_media >= 5.0 and frequencia_media >= 70:
        situacao = "Resultado: APROVADO! Parabéns."

    elif frequencia_media < 70:
        situacao = "Resultado: REPROVADO POR FALTA."

    else:
        situacao = "Resultado: RECUPERAÇÃO (Nota insuficiente)."


    # Juntamos os dados da materia:
    dados_materia = [disciplina, nota_media, frequencia_media, situacao]

    historico.append(dados_materia)

    continuar = input("Deseja cadastrar outra disciplina? (s/n): ")

print(" RESUMO DO HISTÓRICO DO ALUNO ")

# Agora pegamos a pasta e lemos cada papel
for materia in historico:
    # materia[0] é o nome, materia[1] é a nota, materia[2] é a frequência, materia[3] é a situação
    print(f"Disciplina: {materia[0]}")
    print(f"Média: {materia[1]}")
    print(f"Frequência: {materia[2]}%")
    print(f"Status: {materia[3]}")
print("Sistema encerrado.")
```
