len - retorna o número de elementos da lista

```python
num = [10, 4, -23, 3, 12, 25, 6, 5, 7, 8, 90, 10]
len(num)
```

append - adiciona elementos

```python
num = [10, 4, -23]
num.append(5)
num
```

del - remove elementos (começa com 0)

```python
num = [4,5,7,10,11]
del num[0]
num
```

sum - soma elementos

```python
num = [2, 5, 7, 10, 15, 10]
del num[1]
print(num)
sum(num)
```
